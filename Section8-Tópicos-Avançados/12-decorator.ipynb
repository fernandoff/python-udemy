Decorator
- Modificam ou estendem o comportamento de funções existentes de forma reutilizável.
- Adicionam funcionalidades a funções existentes sem modificar seu código diretamente.

```python
def repetir(vezes):
    def decorator(funcao):
        def wrapper(*args, **kwargs):
            for _ in range(vezes):
                resultado = funcao(*args, **kwargs)
            return resultado
        return wrapper
    return decorator

@repetir(1)
def saudar(nome):
    print(f"Olá, {nome}!")

saudar("João")
# Saída:
# Olá, João!
# Olá, João!
# Olá, João!
```

```python
def log_usuario(funcao):
    def wrapper(*args, **kwargs):
        usuario = kwargs.get('usuario', 'Anônimo')
        print(f"Dec - Usuário: {usuario}")
        return funcao(*args, **kwargs)
    return wrapper

@log_usuario
def processar(dados, usuario=None):
    return f"Processando {dados}"

processar("arquivo.txt", usuario="Lala")
```
