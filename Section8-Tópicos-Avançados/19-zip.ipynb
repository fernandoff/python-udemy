Zip
- É combinar múltiplos iteráveis (lista, ou tupla) em um único iterável

```python
from itertools import zip_longest

nomes = ["Alice", "Bob", "Charlie"]
idades = [25, 30, 35]
enderecos = [
    {'cidade': 'Criciúma', 'bairro': 'Santa Bárbara'},
    {'cidade': 'Florianópolis', 'bairro': 'Centro'},
    #{'cidade': 'São Paulo', 'bairro': 'Vila Mariana'}
]
```

```python
for nome, idade, endereco in zip_longest(nomes, idades, enderecos):
    if endereco is None:
        print(f"Nome: {nome}, Idade: {idade}, Cidade: N/A, Bairro: N/A")
    else:
        print(f"Nome: {nome}, Idade: {idade}, Cidade: {endereco.get('cidade', 'N/A')}, Bairro: {endereco.get('bairro', 'N/A')}")
    
```

```python
for nome, idade in zip(nomes, idades):
    print(f"Nome: {nome}, Idade: {idade}")
```

```python
pessoas = dict(zip(nomes, idades))
pessoas
```

```python
pessoas_acima_de_30 = {nome: idade for nome, idade in pessoas.items() if idade > 30}
pessoas_acima_de_30
```
